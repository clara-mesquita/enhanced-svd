```python
#gridsearch
def grid_search_cv(modelo, units, X_train, learning_rates, y_train, epochs_list, batch_sizes, patiences, model_name):
    best_loss = float('inf')
    best_params = {}
    for lr in learning_rates:
        for epochs in epochs_list:
            for batch_size in batch_sizes:
                for patience in patiences:
                    model = modelo(units, X_train, lr)
                    histories = fit_model_with_cross_validation(model, X_train, y_train, model_name, patience, epochs, batch_size)
                    mean_history = calculate_mean_history(histories)
                    val_loss = min(mean_history['val_loss'])
                    print("Val Loss: ", val_loss, "learning rate: ", lr, "epochs: ",  epochs, "batch_size: " , batch_size, "patience: ", patience)
                    if val_loss < best_loss:
                        best_loss = val_loss
                        best_params = {'learning_rate': lr, 'epochs': epochs, 'batch_size': batch_size, 'patience': patience} 
                    del model
                    tf.keras.backend.clear_session()
                    gc.collect()
    print('O modelo '+model_name+ ' tem como melhores parametros os seguintes: learning_rate '+ str(best_params['learning_rate'])+' epochs: '+ str(best_params['epochs'])+' batch_size: '+ str(best_params['batch_size'])+ ' patience: '+ str(best_params['patience']))
    return best_params

#validação cruzada
def fit_model_with_cross_validation(model, xtrain, ytrain, model_name, patience, epochs, batch_size):
    tscv = TimeSeriesSplit(n_splits=5)
    fold = 1
    histories = []
    for train_index, val_index in tscv.split(xtrain):
        x_train_fold, x_val_fold = xtrain[train_index], xtrain[val_index]
        y_train_fold, y_val_fold = ytrain[train_index], ytrain[val_index]
        early_stop = keras.callbacks.EarlyStopping(monitor='val_loss', patience=patience, restore_best_weights=True, min_delta=1e-5)
        history = model.fit(x_train_fold, y_train_fold, epochs=epochs, validation_data=(x_val_fold, y_val_fold), batch_size=batch_size, callbacks=[early_stop], verbose=0) #verbose = 0 para otimizar tempo
        print('\n\nTREINAMENTO - Fold', fold, 'do modelo:', model_name)
        histories.append(history)
        fold += 1   
    return histories 

# calcula a media das metricas obtidas nos historys - validação cruzada
def calculate_mean_history(histories):
    mean_history = {'loss': [], 'root_mean_squared_error': [], 'val_loss': [], 'val_root_mean_squared_error': []}
    for fold_history in histories:
        for key in mean_history.keys():
            mean_history[key].append(fold_history.history[key])
    for key, values in mean_history.items():
        max_len = max(len(val) for val in values)
        for i in range(len(values)):
            if len(values[i]) < max_len: #caso em que nao se treina todas as epocas (patience)
                values[i] += [values[i][-1]] * (max_len - len(values[i])) #completa o restante da lista com o ultimo valor obtido
    for key, values in mean_history.items():
        mean_history[key] = [sum(vals) / len(vals) for vals in zip(*values)]
    
    return mean_history

#função para salvar o modelo
# def save_model(model, directory, substring_desejada, modelo):
#     if not os.path.exists(directory):
#         os.makedirs(directory)
#     file_path = os.path.join(directory, f'{substring_desejada}_{modelo}_final_model.keras')
#     model.save(file_path)
#     print(f"Modelo salvo como '{file_path}'")

#plotar os graficos da media dos treinamentos por epocas: validação cruzada
def plot_loss_cv(mean_history, model_name, link):
    epochs = range(1, len(mean_history['loss']) + 1)
    plt.figure(figsize=(15, 6), dpi=100)
    plt.plot(epochs, mean_history['loss'], label='Train Loss')
    plt.plot(epochs, mean_history['val_loss'], label='Validation Loss')
    plt.title('Mean Training and Validation Loss for '+' '+link + ' '+ model_name)
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.show()

# Main functions

# Create input dataset
# The input shape should be [samples, time steps, features]
def create_dataset (X, look_back = 3):
    Xs, ys = [], []
    
    for i in range(len(X)-look_back):
        v = X[i:i+look_back]
        Xs.append(v)
        ys.append(X[i+look_back])
        
    return np.array(Xs), np.array(ys)

# Create GRU model
def create_gru(units, train, learning_rate): 
    model = Sequential() 
    model.add(GRU(units=units, return_sequences=True, 
                  input_shape=[train.shape[1], train.shape[2]]))
    model.add(GRU(units=units))   
    model.add(Dense(1))
    
    # Garantir que a última camada use float32 para melhor precisão
    model.add(tf.keras.layers.Lambda(lambda x: tf.cast(x, tf.float32)))
    
    model.compile(loss=MeanSquaredError(), 
                  optimizer=Adam(learning_rate=learning_rate), 
                  metrics=[RootMeanSquaredError()])
    return model
# Make prediction
def prediction(model, xtest, ytest, myscaler, model_name, link): 
    prediction = model.predict(xtest) 
    prediction = myscaler.inverse_transform(prediction) 
    dataframe_prediction = pd.DataFrame(data={'Prediction':prediction.flatten(), 'Test':ytest.flatten()})
    save_path = os.path.join('results_gru', 'gru', 'forecast', f'prediction_{model_name}_{link}.csv') 
    os.makedirs(os.path.dirname(save_path), exist_ok=True)
    dataframe_prediction.to_csv(save_path, index=False)
    return prediction

# Plot test data vs prediction
def plot_future(predictionGRU, y_test, link):
    plt.figure(figsize=(15, 6), dpi=100)
    range_future = len(y_test)
    plt.plot(np.arange(range_future), np.array(y_test), label='Test data')
    plt.plot(np.arange(range_future), np.array(predictionGRU), label='GRU')
    
    plt.title('Test data vs prediction for '+ link)
    plt.legend(loc='upper left')
    plt.xlabel('Time')
    plt.ylabel('Throughput (Mbps)')
    
    save_path = os.path.join('results_gru', 'gru', 'plots', link + '.png')
    os.makedirs(os.path.dirname(save_path), exist_ok=True)
    
    try:
        plt.savefig(save_path)
        print(f"A figura foi salva com sucesso em: {save_path}")
    except Exception as e:
        print(f"Erro ao salvar a figura: {e}")
    plt.show()

# Substitua a função grid_search_cv por esta versão com amostragem aleatória
def random_search_cv(modelo, units, X_train, learning_rates, y_train, epochs_list, batch_sizes, patiences, model_name, n_iter=10):
    best_loss = float('inf')
    best_params = {}
    
    # Gerar todas as combinações possíveis
    all_combinations = [(lr, epochs, bs, pat) 
                        for lr in learning_rates 
                        for epochs in epochs_list 
                        for bs in batch_sizes 
                        for pat in patiences]
    
    # Amostrar aleatoriamente um subconjunto
    if len(all_combinations) > n_iter:
        selected_combinations = random.sample(all_combinations, n_iter)
    else:
        selected_combinations = all_combinations
    
    results_file = 'evaluation_rmse_mae.csv'
    file_exists = os.path.isfile(results_file)
    
    for lr, epochs, bs, pat in selected_combinations:
        try:
            model = modelo(units, X_train, lr)
            histories = fit_model_with_cross_validation(model, X_train, y_train, model_name, pat, epochs, bs)
            mean_history = calculate_mean_history(histories)
            val_loss = min(mean_history['val_loss'])
            
            # Salvar resultado parcial
            with open(results_file, 'a', newline='') as f:
                writer = csv.writer(f)
                if not file_exists:
                    writer.writerow(['val_loss', 'learning_rate', 'epochs', 'batch_size', 'patience', 'model_name'])
                    file_exists = True
                writer.writerow([val_loss, lr, epochs, bs, pat, model_name])
            
            print(f"Val Loss: {val_loss:.6f}, LR: {lr}, Epochs: {epochs}, Batch: {bs}, Patience: {pat}")
            
            if val_loss < best_loss:
                best_loss = val_loss
                best_params = {'learning_rate': lr, 'epochs': epochs, 'batch_size': bs, 'patience': pat}
                
        except Exception as e:
            print(f"Erro na combinação: {lr}, {epochs}, {bs}, {pat}. Erro: {str(e)}")
            continue
    
    print(f'Melhores parâmetros para {model_name}: {best_params}')
    return best_params
    
# Calculate MAE and RMSE
def evaluate_prediction(predictions, actual, model_name):
    errors = predictions - actual
    mse = np.square(errors).mean()
    rmse = np.sqrt(mse)
    nrmse = rmse/ ((np.max(actual))-(np.min(actual)))
    mae = np.abs(errors).mean()
    print(model_name + ':')
    print('Mean Absolute Error: {:.4f}'.format(mae))
    print('Root Mean Square Error: {:.4f}'.format(rmse))
    print('Normalized Root Mean Square Error: {:.4f}%'.format(nrmse*100))
    print('')

    return rmse, mae, nrmse, model_name

def bits_para_megabits(df, col_throughput):
    # Dados em MegaBits/s e fill
    df[col_throughput] = df[col_throughput]/1000000
    df[col_throughput] = df[col_throughput].replace(-1, df[col_throughput].mean())
    df[col_throughput] = df[col_throughput].fillna(df[col_throughput].mean())
    return df
                
def visualizacao_series(df, col_throughput, titulo):
    plt.figure(figsize=(18,6))
    df[col_throughput].plot()
    plt.title(titulo)
    plt.ylabel('Throughput (Mbps)')
    plt.legend() 
    plt.show()

# Imports necessários (adicione no início do seu código)
import os
import pandas as pd
import numpy as np
import random
import csv
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import GRU, Dense
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.losses import MeanSquaredError
from tensorflow.keras.metrics import RootMeanSquaredError
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import TimeSeriesSplit
import json
import gc


# Model Training and Prediction
# Defina o diretório raiz onde deseja iniciar a busca (ajuste conforme necessário)
diretorio_raiz = '../bbr_filtrado_imputado'  # ou o caminho onde estão seus datasets
evaluation_results_path = os.path.join('results-test', 'evaluation_rmse_mae.json')
evaluation = {}

# Itere pelos diretórios e subdiretórios
for pasta_raiz, subpastas, arquivos in os.walk(diretorio_raiz):
    for arquivo in arquivos:
        # Verifique se o arquivo é um arquivo CSV e segue o padrão esperado
        if arquivo.endswith('.csv') and '_' in arquivo:
            # Construa o caminho completo para o arquivo
            caminho_arquivo = os.path.join(pasta_raiz, arquivo)
            
            try:
                # Parser do nome do arquivo para extrair protocol_link_model
                nome_arquivo = os.path.splitext(arquivo)[0]  # Remove .csv
                substring_desejada = nome_arquivo  # protocol_link_model
                
                # Carregue o dataset
                df = pd.read_csv(caminho_arquivo)
                
                # Verifique se as colunas esperadas existem
                if 'time' not in df.columns or 'throughput_bps' not in df.columns:
                    print(f"Arquivo {arquivo} não possui as colunas 'time' e 'throughput_bps' esperadas.")
                    continue
                
                # Configure a coluna time como índice
                df['time'] = pd.to_datetime(df['time'])
                df = df.set_index('time')
                df = df.sort_index()  # Ordena por data
                
                # Remove colunas desnecessárias se existirem
                if '0' in df.columns:
                    df = df.drop('0', axis=1)
                
                # Regularizar o dataset para megabits/s
                bits_para_megabits(df, 'throughput_bps')

                # Visualização das séries (opcional - descomente se desejar)
                # visualizacao_series(df, 'throughput_bps', substring_desejada)
                
                print(f'###################### {substring_desejada} ##########################')
                print(f'Média do dataset: {df["throughput_bps"].mean():.4f}')

                # PREDIÇÃO
                tamanho = int(len(df.index) * 0.8)

                # Set random seed para reprodutibilidade
                tf.random.set_seed(7)

                # Split train data and test data
                train_size = tamanho
                train_data = df[:train_size]
                test_data = df[train_size:]
                
                # Preparar dados para o modelo
                train_data = train_data['throughput_bps'].values.reshape(-1, 1)
                test_data = test_data['throughput_bps'].values.reshape(-1, 1)

                scaler = MinMaxScaler().fit(train_data)

                train_scaled = scaler.transform(train_data)
                test_scaled = scaler.transform(test_data)

                X_train, y_train = create_dataset(train_scaled)
                X_test, y_test = create_dataset(test_scaled)

                # Grid search para encontrar os melhores parâmetros
                list_lr = [0.00001, 0.0001, 0.001] 
                list_epochs = [100, 300, 500] 
                list_bs = [32, 64, 128] 
                list_pat = [5, 10]

                # list_lr = [0.00001] 
                # list_epochs = [50] 
                # list_bs = [32] 
                # list_pat = [5]

                # Caminho para salvar o modelo
                caminho_modelo_salvo = 'saved_models'
                
                best_params_gru = grid_search_cv(create_gru, 64, X_train, list_lr, y_train,  list_epochs, list_bs, list_pat, 'GRU')
                # best_params_gru = random_search_cv(create_gru, 64, X_train, list_lr, y_train, 
                #                                list_epochs, list_bs, list_pat, 'GRU')
                
                # Tendo os melhores parâmetros, criar e treinar o modelo
                model_gru = create_gru(64, X_train, best_params_gru['learning_rate'])

                prev_history_gru = fit_model_with_cross_validation(
                    model_gru, X_train, y_train, 'GRU', 
                    best_params_gru['patience'], best_params_gru['epochs'], 
                    best_params_gru['batch_size']
                )
                
                history_gru = calculate_mean_history(prev_history_gru)
                
                # Salvar o modelo
                # save_model(model_gru, caminho_modelo_salvo, substring_desejada, 'GRU')

                # Transform data back to original data space
                y_test = scaler.inverse_transform(y_test)
                y_train = scaler.inverse_transform(y_train)

                # Fazer predições
                prediction_gru = prediction(model_gru, X_test, y_test, scaler, 'GRU', 
                                          link=substring_desejada)

                # Plotar resultados
                #plot_loss_cv(history_gru, 'GRU', substring_desejada)
                plot_future(prediction_gru, y_test, link=substring_desejada)

                # Avaliar predições
                gru_evaluation = evaluate_prediction(prediction_gru, y_test, 'GRU')

                # Guardar valores de avaliação
                chave_gru = f"{substring_desejada}, {gru_evaluation[3]}"
                tupla = (gru_evaluation[0], gru_evaluation[1], gru_evaluation[2])
                evaluation[chave_gru] = tupla

            except pd.errors.EmptyDataError:
                print(f"Arquivo: {arquivo} - O arquivo está vazio, pasta: {pasta_raiz}")
            except Exception as e:
                print(f"Arquivo: {arquivo}, pasta: {pasta_raiz} - Erro: {str(e)}")

# Salvar resultados da avaliação
novo_dicionario = {}
for chave, valores in evaluation.items():
    valor1, valor2, valor3 = valores
    novo_dicionario[chave] = {'RMSE': valor1, 'MAE': valor2, 'NRMSE': valor3}


os.makedirs(os.path.dirname(evaluation_results_path), exist_ok=True)
with open(evaluation_results_path, 'w') as arquivo:
    json.dump(novo_dicionario, arquivo, indent=4)

print("Processamento concluído! Resultados salvos em 'results/gru-imputation/'.")
```
